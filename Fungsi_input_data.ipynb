```python
# Fungsi khusus Anggota A: Input Data Penerimaan Komoditas Agroindustri
def input_data_agroindustri():
    daftar_penerimaan = []
    print("==================================================")
    print("   INPUT DATA PENERIMAAAN BANAN BAKU AGROINDUSTRI  ")
    print("==================================================")

    # Meminta jumlah transaksi/lot yang akan diinput
    while True:
        try:
            jumlah_batch = int(input("Masukkan jumlah lot/batch penerimaan: "))
            if jumlah_batch > 0:
                break
            print("Jumlah batch harus lebih dari 0!")
        except ValueError:
            print("Input tidak valid! Harap masukkan angka bulat.")

    # Loop penginputan data per batch
    for i in range(jumlah_batch):
        print(f"\n--- Transaksi Batch Ke-{i+1} ---")
        pemasok = input("Nama Pemasok / Kelompok Tani : ").strip()
        komoditas = input("Jenis Komoditas (Kopi/Sawit/Susu) : ").strip().title()

        # Validasi input berat/volume bahan baku
        while True:
            try:
                jumlah_bahan = float(input("Jumlah Bahan Baku (kg/Liter)  : "))
                harga_per_unit = float(input("Harga Beli Per Unit (Rp)      : "))
                if jumlah_bahan > 0 and harga_per_unit > 0:
                    break
                print("Jumlah dan harga harus lebih besar dari 0!")
            except ValueError:
                print("Input tidak valid! Harap masukkan angka yang sesuai.")

        # Menyimpan data dalam bentuk dictionary
        batch_data = {
            "id_batch": f"AGRO-{i+1:03d}",
            "pemasok": pemasok,
            "komoditas": komoditas,
            "jumlah_bahan": jumlah_bahan,
            "harga_per_unit": harga_per_unit,
            "total_biaya": jumlah_bahan * harga_per_unit
        }

        daftar_penerimaan.append(batch_data)
        print(f"[SUCCESS] Batch {batch_data['id_batch']} berhasil dicatat!")

    print("\n==================================================")
    print("[INFO] Seluruh data penerimaan berhasil disimpan.")
    return daftar_penerimaan


# Inisialisasi dan pengujian fungsi input
if __name__ == "__main__":
    data_bahan_baku = input_data_agroindustri()
    print("\nData Terstruktur (Siap diproses oleh Anggota B):")
    print(data_bahan_baku)
```
