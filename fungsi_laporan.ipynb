```python
# ==================================================
# FUNGSI KHUSUS ANGGOTA B: LAPORAN PENERIMAAN
# ==================================================

def laporan_penerimaan(daftar_penerimaan):
    print("\n==================================================")
    print("       LAPORAN PENERIMAAN BAHAN BAKU")
    print("==================================================")

    # Mengecek apakah terdapat data penerimaan
    if not daftar_penerima:
        print("Belum ada data penerimaan yang tersedia.")
        return

    total_bahan = 0
    total_biaya = 0

    # Menampilkan data setiap batch
    print(f"{'ID Batch':<12} {'Pemasok':<20} {'Komoditas':<12} "
          f"{'Jumlah':>12} {'Total Biaya':>18}")
    print("-" * 78)

    for batch in daftar_penerimaan:
        print(f"{batch['id_batch']:<12} "
              f"{batch['pemasok']:<20} "
              f"{batch['komoditas']:<12} "
              f"{batch['jumlah_bahan']:>12.2f} "
              f"Rp {batch['total_biaya']:>15,.0f}")

        total_bahan += batch["jumlah_bahan"]
        total_biaya += batch["total_biaya"]

    print("-" * 78)
    print(f"Total jumlah bahan baku : {total_bahan:,.2f}")
    print(f"Total biaya pembelian   : Rp {total_biaya:,.0f}")
    print(f"Jumlah batch             : {len(daftar_penerimaan)}")

    print("==================================================")


# ==================================================
# PENGUJIAN FUNGSI LAPORAN
# ==================================================

if __name__ == "__main__":
    data_bahan_baku = [
        {
            "id_batch": "AGRO-001",
            "pemasok": "Kelompok Tani Makmur",
            "komoditas": "Kopi",
            "jumlah_bahan": 100,
            "harga_per_unit": 25000,
            "total_biaya": 2500000
        },
        {
            "id_batch": "AGRO-002",
            "pemasok": "Kelompok Tani Sejahtera",
            "komoditas": "Susu",
            "jumlah_bahan": 150,
            "harga_per_unit": 8000,
            "total_biaya": 1200000
        }
    ]

    laporan_penerimaan(data_bahan_baku)
```
